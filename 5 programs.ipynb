find the factorial of a number

```python
import math
num = int(input("Enter a number"))
result = math.factorial(num)
print(f"The factorial of a {num} is {result}")
```

```python
n = int(input("Enter number:"))
def get_factorial(n):
  fact = 1
  for i in range(1,n+1):
    fact*=i
    return fact
print(get_factorial(n))

```

```python
n= int(input("Enter a number"))
def get_factorial(n):
  if n==0 or n==1:
    return 1
  return n*get_factorial(n-1)
print(get_factorial(n))
```

```python
n = int(input("Enter a number"))
reversed_num = 0
while n>0:
  digit = n%10
  reversed_num = (reversed_num*10)+digit
  n=n//10
print(reversed_num)
```

write a function to check a number is prime

```python
n = int(input("Enter number: "))
if(n < 2):
    print("Not Prime")
if(n == 2):
    print("prime")
for i in range(2,n//2):
    if(n%i == 0):
        print("Not Prime")
        break
    else:
        print("Prime")
```
